# feature Scaling

```python
from sklearn.preprocessing import StandardScaler

cols = ['age', 'bmi', 'children']
```

```python
scaler = StandardScaler()
```

```python
df_cleaned[cols] = scaler.fit_transform(df_cleaned[cols])
```

```python
df_cleaned
```

```python
from scipy.stats import pearsonr

selected_features = ['age', 'bmi', 'children', 'is_female', 'is_smoker', 'region_northwest', 'region_southeast', 'region_southwest','bmi_category_Normal weight', 'bmi_category_Overweight', 'bmi_category_Obesity']


correlation={
    feature: pearsonr(df_cleaned[feature], df_cleaned['charges'])[0] for feature in selected_features
}

correlation_df = pd.DataFrame(list(correlation.items()), columns=['Feature', 'Pearson Correlation'])
correlation_df = correlation_df.sort_values(by='Pearson Correlation', ascending=False)
```

```python
correlation_df
```

```python
cat_features = ['is_female', 'is_smoker', 'region_northwest', 'region_southeast', 'region_southwest','bmi_category_Normal weight', 'bmi_category_Overweight', 'bmi_category_Obesity']
```

```python
from scipy.stats import chi2_contingency
```

```python
alpha = 0.05
```

```python
df_cleaned['charges_bin'] = pd.qcut(df_cleaned['charges'], q=4, labels=False)
```

```python
chi2_results = {}
for feature in cat_features:
    contingency_table = pd.crosstab(df_cleaned[feature], df_cleaned['charges_bin'])
    chi2, p, _, expected = chi2_contingency(contingency_table)
    decision = "Reject Null Hypothesis" if p < alpha else "Fail to Reject Null Hypothesis"
    chi2_results[feature] = {
        'Chi2 Statistic': chi2,
        'p-value': p,
        'Decision': decision
    }
```

```python
chi2_df = pd.DataFrame(chi2_results).T
```

```python
chi2_df = chi2_df.sort_values(by='p-value', ascending=True)
```

```python
chi2_df
```
